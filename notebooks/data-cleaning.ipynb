```python
import sqlite3
import pandas as pd

sql_query = """
SELECT
    p.age,
    p."capital-gain" AS capital_gain,
    p."capital-loss" AS capital_loss,
    p."hours-per-week" AS hours_per_week,
    p.income,
    o.workclass,
    o.occupation,
    d."marital-status" AS marital_status,
    d.relationship,
    d.race,
    d.sex,
    d."native-country" AS native_country,
    e.education
FROM person AS p
LEFT JOIN occupation AS o ON p.occupation_id = o.id
LEFT JOIN demographics AS d ON p.demographics_id = d.id
LEFT JOIN education AS e ON p.education_id = e.id
"""

with sqlite3.connect("../data/census_data.db") as connection:
    df = pd.read_sql_query(sql_query, connection)

df
```
